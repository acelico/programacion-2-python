2. Write a program that calculates the factorial of a certain number introduced using input user. The program will have to check if the number introduced is an integer.

```python
num1= input('Introduce an integer numeric value to calculate its factorial: ')
factorial=1
try:
    num2 = int(num1)
    for i in range(1,num2+1):
      factorial=factorial*i
    print('The factorial of %d is %d'%(num2,factorial))
except:
    print('I cannot calculate the factorial of a non-integer number')
```

5. Write a program that reads a sequence of characters that ends in a point ('.'), and tells us how many times the letter 'a' appears without taking care if it is in lower or upper case.

~~~


```python
ch = input('Introduce one character: ')
cont=0
while ch !='.':
    if ch=='A' or ch =='a':
        cont+=1
    ch = input('Introduce another character: ')
print('There are {} as'.format(cont))
```

```python
import csv
csv_path = "PPR-2015.csv"
list_h=[]

#The context manager opens the file for reading in a variable file
with open(csv_path, "r") as file:
    reader = csv.reader(file)
    # we can use a for-in loop
    for n, row in enumerate(reader):
        list_h.append(row)
        # print(row)
```

```python
l_prices=[]
num_col=4 # column of the Price in euros
for row in range(1, len(list_h)):
    date=list_h[row][0] #We extract the date
    if date[3:5]=='01':
          price=list_h[row][num_col] #We extract the price of the house, with the format '€138,000.00'
          price=price.split('.') #We split the string in two strings: ['€138,000','00']
          price=price[0]#We get the first part
          l_prices.append(int(price[1:].replace(',', ''))) #extract the characters starting in the second and replace the',' by nothing, and finally convert it into a number

av_price=0
for price in l_prices:
  av_price+=price
av_price=av_price/len(l_prices)
print('The average price of the houses sold in January in Dublin is %f'%av_price,' euros')
```

9. Using the file 'PPR-2015.csv', write a program that calculates how many different postal codes there are in the file and print them.

```python
l_postal_codes=[]
num_col=2 # column of the postal code
num_dif_pc=0 #variable to count different postal codes
for row in range(1, len(list_h)):
    pc=list_h[row][num_col] #We extract the postal code
    if pc!='':
       if not(pc in l_postal_codes):
          num_dif_pc+=1
          l_postal_codes.append(pc)
print(l_postal_codes)
print('The number of different postal codes in the file is %d'%num_dif_pc)
```
