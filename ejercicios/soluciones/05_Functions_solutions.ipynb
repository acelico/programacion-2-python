1. Create a function named *good_job_order* that, given two input variables: a list of names, and a variable to choose ordering the names in alphabetical order or reverse alphabetical order, print the following messages (if list_names=['lola','patxi','nerea','xabier']):

~~~
The names are in alphabetical order: (or The names are in reverse alphabetical order:)

Good job, Lola!
Your performance in this subject has been amazing.

Good job, Nerea!
Your performance in this subject has been amazing.

Good job, Patxi!
Your performance in this subject has been amazing.

Good job, Xabier!
Your performance in this subject has been amazing.
~~~




```python
def good_job_order(list_names,order):
    #order is a binary variable: if equal to 1, the names are listed in order; if not, inverse alphabetical order
    good_job = ["\nGood job, %s!"%name.title() for name in list_names]

    if order==1:
          print("The names are in alphabetical order:")
          for g_job in sorted(good_job):
             print(g_job)
             print("Your performance in this subject has been amazing.\n")
    else:
      print("The names are in reverse alphabetical order:\n")
      for g_job in sorted(good_job,reverse=True):
          print(g_job)
          print("Your performance in this subject has been amazing.\n")
```

```python
list_names=['lola','patxi','nerea','xabier']
order=1
good_job_order(list_names,order)
order=0
good_job_order(list_names,order)
```

```python
def sum_calculator_3(num1, num2, *nums):
    sum = num1+num2
    msg="The sum of %f, %f, "%(num1,num2)
    for num in nums:
        sum+=num
        msg+="%f, "%num
    print(msg+"is: %f"%sum)
```

7. Create a function called *most_common_char* that takes a string as a argument and returns the most common character in that string.

```python
def most_common_char(string):
    max_count=0
    most_com_ch=''
    for index,ch in enumerate(string):
      if ch!=' ':
        count=0
        for index2,ch2 in enumerate(string):
           if ch2==ch:
              count+=1
        if count>max_count:
           max_count=count
           most_com_ch=ch
    return most_com_ch
```

```python
ch='A'
print(ch.isalpha())
print(ch.isdigit())
```

```python
def print_digit_letter(string):
    count_letter=0
    count_digit=0
    for ch in string:
      if ch.isalpha():
        count_letter+=1
      elif ch.isdigit():
        count_digit+=1
    print("The number of letters in this string is: %d"%count_letter)
    print("The number of digits in this string is: %d"%count_digit)
```

9. Write a Python function called *word_frequency* that takes a list of words as input and returns a dictionary where the keys are unique words and the values are the frequency of each word in the list.

```python
def word_frequency(word_list):
    dict = {}
    for word in word_list:
        if word in dict:
            dict[word] += 1
        else:
            dict[word] = 1
    return dict
```

```python
word_list=['coche','moto','paz','alegría','esperanza', 'paz']
word_frequency(word_list)
```

10. Write a function called *is_valid_email*  that takes an email address as an argument and returns True/False depending on whether it is a valid email address.Check rules for a valid email:
 - Must contain at least 1 character before the at symbol
 - Must contain an @ symbol
 - Must have at-least 1 character after the @ symbol and before the period(.)
 - Must contain at least 1 character after the last period(.).
 - Maximum 256 characters
 - Must start with a letter or a number

```python
#Supposing that the email address is a string
def is_valid_email(email_address):
  if email_address[0].isalpha() or email_address[0].isdigit():#Must start with a letter or a number
     if len(email_address)<=256: #Maximum 256 characters
        if '@' in email_address:# Must contain an @ symbol
          pos_at=email_address.index('@')
          before_at=email_address[0:pos_at]#part of the email before @
          is_ch=0
          for i in range(0,len(before_at)): #Must have at-least 1 character before the @ symbol
              if before_at[i].isalpha():
                is_ch=1
          if is_ch==1: # It has at-least 1 character before the @ symbol
             if '.' in email_address[pos_at+1:]: #if there is a point after the @ symbol
                after_at=email_address[pos_at+1:]
                pos_point=after_at.index('.') #position of the point
                is_ch=0
                for i in range(1,pos_point): #Must have at-least 1 character after the @ symbol and before the period(.)
                   if after_at[i].isalpha():
                       is_ch=1
                if is_ch==1: #It has at-least 1 character after the @ symbol and before the period(.)
                   after_point=after_at[pos_point:] #last part after the point
                   is_ch=0
                   for i in range(0,len(after_point)): #Must have at-least 1 character after the @ symbol and before the period(.)
                      if after_point[i].isalpha():
                         is_ch=1
                   if is_ch==1:
                      return True
                   else:
                      return False
                else:
                   return False
             else:
                 return False
          else:
            return False
        else:
          return False
     else:
      return False
  else:
    return False
```
