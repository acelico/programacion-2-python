4. Write a program that prompts the user to enter a sentence via keyboard input and generates a dictionary with the count of occurrences of each word in the sentence.
For example, if it receives "What a lovely day it is today" it should return:

- 'what': 1, 'a': 1, 'lovely': 1, 'day': 1, 'it': 1, 'is': 1, 'today': 1

```python
sentence = input('Introduce a sentence: ')
#type(sentence)   #if you execute this commands you will see sentence is a list of characters
#for word in sentence:
#  print(word,'\n')
```

```python
i=0 #index to go through all the characters of sentence
dic_words={} #dictionary with the words and their occurrences, initializing to empty dictionary
word='' #auxiliar variable to write the words in the sentence
while i <len(sentence): #While we do not get until the end of the sentence
     if sentence[i]==' ': #if the character in the position i is a blank space
        if len(word)>0:   #and if also the length of variable word is greater than 0 (we have been inserting characters before forming a word)
           if word in dic_words.keys(): #if the word in variable word is already in the dictionary
              dic_words[word]=dic_words[word]+1 #add 1 to the occurrences
           else: #if the word is not in the dictionary
              dic_words[word]=1 #create the new word in the dictionary and add ocurrence=1
              word='' #initialize the word again to capture a new one
        if i<len(sentence):
           i+=1 #increment the index to look for another word in sentence
        else:
          break
     else: #sentence[i] is a character
        word+=sentence[i] #we add the character to the variable word
        if i<len(sentence):
           i+=1 #increment the index to look for another word in sentence
           if i==len(sentence):
              if word in dic_words.keys(): #if the word in variable word is already in the dictionary
                 dic_words[word]=dic_words[word]+1 #add 1 to the occurrences
              else: #if the word is not in the dictionary
                 dic_words[word]=1 #create the new word in the dictionary and add ocurrence=1
                 word='' #initialize the word again to capture a new one
```

```python
for word, value in dic_words.items():
    print('\n Word: %s' % word)
    print('Occurrences: %s' % value)
```
