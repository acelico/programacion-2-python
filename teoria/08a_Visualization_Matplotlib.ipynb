## Axes and Figures coordinate systems
If you need to "position" things inside your graph you should use coordinate systems. Axes and Figures use different coordinates systems:

- Axes use the same coordinate system as the data (i.e., data coordinates).
- Figure coordinates range from 0 to 1 on both the x and y axes.(being 0.5 the center)

You can use them jointly in your graphs!

````python
#axes coordinate system
fig, ax = plt.subplots()

ax.plot([1,2,3,4,5], [2,3,1,4,5])
ax.text(3, 3, "Hello world!") # use axes coordinates 

plt.show()
````

````python
# Figures coordinate system
fig, ax = plt.subplots()

ax.plot([1,2,3,4,5], [2,3,1,4,5])
fig.text(0.5, 0.5, "Hello world!") # use figures coordinates

plt.show()
````

### Resume
- Every Matplotlib visualization is a Figure with one or more Axes.
- A Figure is the overall container for all elements, including Axes and other graph components. Its coordinates range from 0 to 1 on both x and y axes.
- An Axes is a container for displaying a single chart (line chart, bar plot, etc.), with coordinates in the same scale as the data.
- When a Figure has a single Axes (as in every chart we've made so far), the center of the Axes and the Figure align.

## Subplots and plots combinations
Matplot lib allows to combine multiple plots by its *figure()* and *subplots()* functions.

### An example with categorical data
Let's say we have data about 3 groups (A,B,C). We want to plot the same data with different visualization solutions.

````python
#The example
names = ['group_a', 'group_b', 'group_c']
values = [1, 10, 100]
# First we create a figure with a width of 9 inches and a height of 3 inches
plt.figure(figsize=(9, 3)) #you can modify this according to your needs!
#Now create the plots
plt.subplot(131) # Create the first subplot (1 row, 3 columns, position 1)
plt.bar(names, values)
plt.subplot(132) # Create the first subplot (1 row, 3 columns, position 2)
plt.scatter(names, values)
plt.subplot(133) # Create the first subplot (1 row, 3 columns, position 3)
plt.plot(names, values)
plt.suptitle('Categorical Plotting')
plt.show()
````

````python
#Same example: different orientations
names = ['group_a', 'group_b', 'group_c']
values = [1, 10, 100]
# First we create a figure with a width of 4 inches and a height of 12 inches
plt.figure(figsize=(4, 12)) #This is now larger!
#Now create the plots
plt.subplot(311) # Create the first subplot (3 rows, 1 columns, position 1)
plt.bar(names, values)
plt.subplot(312) # Create the first subplot (3 rows, 1 columns, position 2)
plt.scatter(names, values)
plt.subplot(313) # Create the first subplot (3 rows, 1 columns, position 3)
plt.plot(names, values)
plt.suptitle('Categorical Plotting')
plt.show()
````

**Heatmap** <br>
A heatmap is a type of data visualization that uses colors to represent numerical values in a matrix format. It helps to quickly identify patterns, trends, and outliers in a dataset.<br>

Darker or warmer colors (e.g., red) often indicate higher values.
Lighter or cooler colors (e.g., blue) often indicate lower values. <br>

For example, say I want tovisualize the test scores of 4 students across 4 subjects...

````python
import numpy as np

#Students and subjects
subjects = ["Math", "Programming", "Econometrics", "Law"]
students = ["Bob", "Charlie", "Julia", "Ana"]

# Scores (should be a 2D array of numbers)
scores = np.array([
    [10, 9, 8, 7],  # Bob
    [7, 8, 9, 6],   # Charlie
    [9, 7, 6, 8],   # Julia
    [6, 5, 7, 10]   # Ana
])

# Create figure
fig, ax = plt.subplots()

# Display heatmap
cax = ax.imshow(scores, cmap="coolwarm")

# Add labels
ax.set_xticks(np.arange(len(subjects)))
ax.set_yticks(np.arange(len(students)))
ax.set_xticklabels(subjects)
ax.set_yticklabels(students)

# Add colorbar
plt.colorbar(cax, label="Score")

# Show plot
plt.show()
````

````python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 4, 2]

fig, ax = plt.subplots(figsize=(4,3))
ax.plot(
  x, y,
  color="lightgrey",
  linewidth=2,
  zorder=1
)
ax.scatter(
  x, y,
  color="#69b3a2",
  linewidth=1,
  edgecolor="black",
  zorder=1
)
plt.show()
````

# 1) What can you do?
You have almost infinite options when it comes to choosing the right visualization solution...that does not mean that every tool is the appropriate one. <br>

Next goes an example:<br>
Sales for different product categories. <br>
Multiple visualization solutions can be used...

````python
# Toy data
# Example dataset: Sales performance of different product categories
import squarify

categories = [
    "Electronics", "Clothing", "Home & Kitchen", "Books", "Toys", "Beauty", "Sports", "Automotive", "Food & Beverages", "Health"
]
sales = [500, 400, 350, 300, 280, 260, 240, 220, 200, 180]  # Sales in thousands

# --- TREEMAP ---
fig, ax = plt.subplots(figsize=(10, 6))
colors = plt.cm.tab10.colors[:len(categories)]  # Assign different colors
squarify.plot(sizes=sales, label=categories, color=colors, alpha=0.7, text_kwargs={'fontsize': 10}, linewidth=2, edgecolor='white')
plt.axis('off')
plt.title("Treemap of Product Category Sales", fontsize=14)
plt.show()


# --- BAR CHART ---
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(categories[::-1], sales[::-1], color='skyblue')
ax.set_xlabel("Sales (in thousands)")
ax.set_title("Bar Chart of Product Category Sales")
plt.show()

# --- PIE CHART ---
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(sales, labels=categories, autopct='%1.1f%%', colors=plt.cm.Paired.colors)
ax.set_title("Pie Chart of Product Category Sales")
plt.show()
````

## The question

````python
import matplotlib.pyplot as plt
import numpy as np

# Toy data
countries = [
    "United States", "Russia", "France", "Germany (FRG)", "Israel", "United Kingdom", "Netherlands", "China", "Spain", "Italy",
    "South Korea", "Turkey", "Ukraine", "Switzerland", "Norway", "Czech Republic", "Indonesia", "Australia", "Canada", "Sweden"
]

values = [100, 40, 35, 30, 25, 22, 20, 18, 16, 14, 13, 12, 11, 10, 9, 8, 7, 6, 5, 4]

# Sorting for better visualization
sorted_indices = np.argsort(values)[::-1]
countries = np.array(countries)[sorted_indices]
values = np.array(values)[sorted_indices]

# Create horizontal bar chart
fig, ax = plt.subplots(figsize=(8, 6))
ax.barh(countries, values, color='teal')
ax.invert_yaxis()  # Invert y-axis to match the original image order
ax.set_xlabel("Value")
ax.set_title("Some fake data about countries")
ax.text(50, 10, "What questions can this graph answer?", color='red', fontweight='bold')
ax.text(28, 15, "What questions  this graph SHOULD NOT try to answer?", color='red', fontweight='bold')


plt.show()
````

**It should answer:**<br>
- Which is the highest (lower) country value?
- Is China doing better thab Russia?

**It should not answer:**
- What percentage of the total is represented by Australia?

## Exploratory vs explanatory graphs

Some graphs are very useful to investigate data, yet they are not suited to be presented to your reader.<br>
A correlogram is an example: it contains lots of information but your user can very easily get lost


````python
# library & dataset
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('iris')

# with regression
sns.pairplot(df, kind="reg")
plt.show()

# without regression
sns.pairplot(df, kind="scatter")
plt.show()
````
