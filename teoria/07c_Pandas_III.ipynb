We are going to charge a dataset related to the Human Development Index (HDI).HDI is a composite index measuring average achievement in three basic dimensions of human development-a long and healthy life, knowledge and a decent standard of living. The HDI is the geometric mean of normalized indices for each of the three dimensions.

Indicators- Life expectancy at birth, Mean Years of Schooling, Gross National Income Per Capita.

```python
df_hdi= pd.read_csv('HDI.csv')
df_hdi.head()
```

```python
df_hdi.info()
```

## The method `.melt`
Pandas melt() function is useful to change a DataFrame into a format where one or more columns are identifier variables (columns that are not alterated), while all other columns, considered measured variables, are unpivoted to the row axis, leaving just two non-identifier columns, variable and value.

In this sense, we are going to leave the clumns HDI Rank and Country and we are going to convert the years into columns with their corresponding values.

```python
df_hdi= pd.melt(df_hdi, id_vars=["HDI Rank", "Country"],#identifiers columns 
                  var_name="Year", value_name="HDI value")
df_hdi.head()
```

## The method `pivot_table`

This method works very similarly to Excel's pivot_table, allowing us to perform operations collectively on a DataFrame. Let's see how it works. Suppose we have a new data file called foods.csv that indicates the name, gender, city, frequency, food, and amount spent by various individuals and cities. So let's load our new DataFrame.

```python
df_foods = pd.read_csv("foods.csv")
df_foods.head()
```

```python
df_foods.info()
```

```python
print(df_foods["City"].unique())
```

Imagine we want to obtain the mean of Spend per Gender and City:


```python
df_foods.pivot_table(values="Spend", index=["Gender","City"], aggfunc= ["count","sum"])
```

 The first argument, `values`, specifies which value we want to pivot on, the second specifies which index we are going to use to display the information, and `aggfunc` specifies the aggregation function we want to apply. We can create a MultiIndex by supplying a list to the index parameter. Let's see how it works. Suppose we want to obtain the sum of the purchased foods indexed by gender and also want to see in which city.

```python
df_foods.pivot_table(values = "Spend", index = ["Gender", "Item"],
                  columns = "City", aggfunc = "sum")
```

It is really interesting as we have obtained a MultiIndex DataFrame with indexes Gender and Item but also with the values added by City as well. We can also obtain a MultiIndex in columns if we add a list of values, as it happens next.

```python
df_foods.pivot_table(values = "Spend", index = ["Gender", "Item"],
                  columns = ["Frequency", "City"], aggfunc = "max").head()
```
