## Indexing using loc, iloc


### .loc()
The .loc method in pandas is a **label-based** data selection and slicing method used to access specific rows and columns in a dataframe. It allows selection based on index labels rather than numerical positions.

Syntax:


```
df.loc[row_labels, column_labels]
```

````python
#First, call a dataframe
#you can declare the index directly when reading the .csv
df_mov = pd.read_csv("movies.csv", index_col = "original_title")
# we order the movies alphabetically
df_mov.sort_index(inplace = True)
df_mov
````

````python
'Django Unchained' in df_mov.index
````

## Method  `.apply()`
The .apply() method in pandas allows you to apply a function to *each element*, row, or column of a DataFrame or Series.<br>

For *DataFrames* there is a method called *apply* that allows us to change the content of the rows. It "applies" a user-described function to an object.<br>

- It is basically a simplified Python loop

````python
#Let's define a function assigning a category to movies based on their score
# This is a function that requires a row element as parameter
def caract_movie(row):

    avg_vote = row['avg_vote'] # takes the value from column avg_vote

    if avg_vote>=9 :
        return "Excellent movie"
    elif avg_vote>=8.5 :
        return "Very good movie"
    else:
        return "Good movie"
#We could define a function that requires columns element, then .apply()would not be needed and we could apply the function directly
   
   # def caract_movie(avg_vote):
   #  if avg_vote >= 9:
   #      return "Excellent movie"
   #  elif avg_vote >= 8.5:
   #      return "Very good movie"
   #  else:
   #      return "Good movie"
````

````python
# We call apply row wise ( or column by column)
df_mov['Characterization'] = df_mov.apply(caract_movie, axis = "columns")#.apply() applies the caract_movie() (iterates) to each row of df_mov because we set axis="columns" (equivalent to axis=1).
#The function is applied row-wise, meaning it evaluates each movie's rating (avg_vote) and assigns a classification.
df_mov.head()
````

## Grouping (split-apply-combine) using `.groupby()`

One of the great strengths of Pandas lies in its simplicity when it comes to splitting, applying functions, and combining again. This is known as the split-apply-combine methodology.The following DataFrame is billionaries.csv. Building off the Forbes World’s Billionaires lists from 1996-2014, scholars at Peterson Institute for International Economics have added a couple dozen more variables about each billionaire - including whether they were self-made or inherited their wealth. (Roughly half of European billionaires and one-third of U.S. billionaires got a significant financial boost from family, the authors estimate.)

````python
df_bill = pd.read_csv("datos/billionaires.csv",index_col = "rank")
df_bill.head(10)
````

````python
# We are going to group by sectors, we create a groupby object
sectors = df_bill.groupby("company.sector")
type(sectors)
````

Now we can apply mathematical functions to get insights of the data.

For example, what are the five sectors with the highest mean wealth?

````python
sectors['wealth.worth in billions'].mean().sort_values(ascending=False).head()
````

We can group by more than one column.<br>
For example we can get to know what is the average wealth of each sector in a given year

````python
sectors = df_bill.groupby(['company.sector','year'])
sectors['wealth.worth in billions'].mean()
````

Differently from R, in pandas it is not necessary to ungroup() dataframes.

### Method `.agg()`

````python
df_bill= pd.read_csv("billionaires.csv", index_col = "rank")
sectors = df_bill.groupby("company.sector")
df_bill.head(3)
````

The method _agg_ allows us to apply different calculus simultaneously.

````python
# I want to see the min, max and mean of the wealth.worth in billions

sectors.agg({"wealth.worth in billions" : ["min", "max","mean"]})
````

An we can use _apply_ for the different groups. Imagine we want to calculate the sectors with highest value in billions. Using _groupby_ we can define a function called _ranker_ to label the sectors from 1 to N. We call *apply* to apply the function to the groups

````python
def ranker(df):
    """Assigns a position in the ranking to each company according to its profit, with 1 being the highest profit generator. Assume that the data is sorted in descending order.."""
    df['sector_worth_rank'] = np.arange(len(df)) + 1
    return df
````

````python
df_bill = pd.read_csv("billionaires.csv")
df_bill = df_bill.sort_values('wealth.worth in billions', ascending=False)
df_bill.head()
````

````python
# We create a new column using the function
df_bill = df_bill.groupby('company.sector').apply(ranker)
# Sectors with more value in billions
df_bill[df_bill['sector_worth_rank'] == 1].head(5)
````
