## Object oriented programming (OOP)
Python is an object-oriented programming (OOP) language. This means that code is organized into objects that combine data **(attributes)** and functionality **(methods)**.<br>

**Data**: Stored as attributes (values).<br>
**Functionality**: Defined as methods (actions the object can perform).<br>
For example, when you create a variable, it becomes an object in Python.<br>

Python provides several built-in data types (e.g., integers, strings, floats), and these are objects too. Each object:

1.   Has **attributes** (data it stores).
2.   Has **methods** (functions you can call to perform actions).


**F-strings** <br>
F-strings or .format substitute values from stored objects into a string. <br>
Alternatively, an f-string is a string (text) that can include variables. You write an f before the quotes and put variables inside {}. <br>
F-strings are the evolution of the .format command. Let's start from .format

```python
#Format command
#Let's answer this question usign python:
#What is the square of 8? And its cube?
print("The result is: {0} {1} and {2} {3}".format('1) The square of 8 is', 50*2, '2) the cube of 8 is', 8*8*8))
```

Explanation: <br>
The .format() method replaces the placeholders {0}, {1}, {2}, and {3} in the string with the corresponding values provided as arguments. Here:

{0} gets replaced by '1) The square of 8 is'<br>
{1} gets replaced by 8 * 8 (which is 64)<br>
{2} gets replaced by '2) The cube of 8 is'<br>
{3} gets replaced by 8* 8 *8 (which is 512)<br>

## Variables casting
Variable casting in programming refers to converting a variable from one *data type to another*. In Python, you can change the type of a variable using built-in functions, and this process is known as type casting.

Python is a dynamically-typed language, meaning variables do not have a fixed type, and you can easily change the type of a variable.

To do so we use functions:
- str()
- int()
- float()

```python
#Example: variables casting

#1) From float to int
#Define a variable float x
x = 5.0
print(type(x))
#Convert x to integer
x = int(x)
#Check its type after the casting
print(type(x))
```

```python
#Example: variables casting

#2) From float to string
#Define a variable float x
x = 5.0
print(type(x))
#Convert x to string
x = str(x)
print(type(x))
```

Obviously, strings cannot be converted into float or integers:

```python
#Example: variables casting - can't convert string into integers
int("hello")
```

```python
#Careful: what seems like a number might not be such!
year = "2025"
print(year)
print(type(year))
# in this case a string can be converted to integer or float
year = int(year) # we must overwrite the previous object
print(type(year))
```

**Why use casting?** <br>
Mostly  for *compatibility* reasons: When working with different data types, sometimes it is necessary to convert them so that operations can be performed without errors (e.g., adding a number to a string).

### Logical operators
Logical operators re used to compare objects and return a boolean (True/ False as a result).<br>
![](https://drive.google.com/uc?id=1SAV5pe1iyN8IpTHZ6o1VvPpp4RiOsUf9)

**and**: Returns True if both conditions are True.<br>
**or**: Returns True if at least one condition is True.<br>
**not**: Reverses the boolean value (negation).<br>


```python
#Examples: AND operator
a = 10
b = 20

# Check if a is less than 15 AND b is greater than 15
result = a < 15 and b > 15
print("Result of `AND` condition is: ", result)  # True (both conditions are True)
#show other cases
```

```python
#Examples: OR operator
a = 10
b = 50

# Check if a is greater than 15 OR b is less than 15
result = a > 15 or b < 15
print("Result of `or` operator is: ", result)  # True (because b < 15 is True)
```

```python
#Examples: NOT operator
a = 10

# Check if NOT (a is greater than 15)
result = not a > 15
print("Result of 'not' operator: ", result)  # True (a > 15 is False, so `not` makes it True)
```

```python
#Example: logical operators
necessity_to_hire = True
available_budget = 35000
hiring_cost = 29000
#specfy the condition
justifiable_hiring = necessity_to_hire and hiring_cost <= available_budget

print("Is the new employee hiring justified?", justifiable_hiring,"!", "Because cost is lower than the available budget.")
```

# Priority of operators

If several expressions are combined together, Python applies priority rules to know the order of execution of the operations.

- ** (exponentiation) :priority 1
- +,- (change of sign): priority 2
- *, /, %, // (arithmetic operators) : priority 3
- +,- (addition and subtraction): priority 4
- <, <=, >, >=, !=, ==: priority 5
- not, or, and: priority 6


Let's see some examples:

```python
x = 2
y = 10
z = 15
t = 5


v = (x + y) * z / t       #( 12 * 15 ) / 5
print(" (x + y) * z / t=",  v)

v = (x + y) * (z / t)     # (12 * 3 )=36
print("((x + y) * z) / t es ",  v)
```
