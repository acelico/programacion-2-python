# **04. Dictionaries in Python**

Dictionaries in Python are a type of **data structure** that allows storing an **unordered** set of **key-value pairs**, with keys being **unique** within the same dictionary (meaning that two elements with the same key cannot exist).

Dictionaries can store **any type of data**.

Dictionaries do not store their information in any particular order, so it is possible that you may not retrieve your information in the same order in which you entered it.

**Syntax**:
~~~
A dictionary is defined using curly braces {} with key-value pairs separated by colons :.

student = {
    "name": "Alice",
    "age": 25,
    "major": "Economics"
}

Key: name --> Value: Alice
~~~

A dictionary admits as values:
- single values (strings, float, integers, boolean)
- lists
- dictionaries

Dictionary are particularly useful for storing non-tabular data thanks to their lack of a pre-determined shape. <br>
Dictionaries are different from dataframes! Dataframes are a 2-dimensional table organised in rows and columns, dictionaries allow for multiple structures other than rows and columns.<br>

**Why don't we have dictionaries in R ?** <br>
Python is a language for general purpose programming, and dictionaries are a flexible instrument for general purpose programming. <br>
R was designed very differently. From day one, R was built for statistics and vectors, not general-purpose programming. 

```python
#Example: creating a dictionary
dic = {'name' : 'María',
       'age' : 19,
       'courses': ['Programming II','Econometrics I'] }
print(dic)
```

## Adding & Modifying Elements in a dictionary
Dictionaries are **mutable**, so you can add or modify key-value pairs.<br>
**Logic**:
~~~
If a key exists and we want to modify its value:
dict["existing_key"] = "new_value" modifies the previous value for the existing key with the new one

If the key does not exist, and we want to create a new one:
dict["new_key"]="new_value" adds a new key

~~~

```python
#Example: adding elements to an empty dictionary
# We can use different ways to start an empty dictionary
dic = {} # call an empty ditcionary equivalent to dic = dict()
dic['name'] = 'María'
dic['age'] = 20
dic['courses'] = ['Programming II','Econometrics I']
print(dic)
# the dictionary will contain all new key:value pairs
```

We can access to one element of the dictionary thorugh the the keys of the dictionary:

```python
 #Example: accessing a key returns its values
print(dic['name'])
print(dic['age'])#unique values
print(dic['courses'])#multiple values
```

If we have more than one element in one key we can access them by using indexing _[ ]_:

```python
#Example: using indexing to identify specific values within a multi-value dictionary
print(dic['courses'][0])#access the first value of key "courses"
print(dic['courses'][1])
```

If we try to access keys that do not exist in the dictionary, Python will throw an error.

```python
print(dic['height']) # height does not exist
```

## Modifying keys' name in a dictionary

Modifying a value in a dictionary is simple because it only depends on the value. Modifying a key is a bit more difficult because each key is **used to unlock one and only one value**. We can change a key in two steps:

- Create a new key and copy the value to the new key.
- Delete the old key, which also removes the old value.

Here's an example. We'll use a dictionary with just one key-value pair to keep things simple.

```python
# Imaging we want to translate the keys into spanish:

person['nombre'] = person['name'] #First create a new key and copy the existing values to the new key
print(person) # now he dictionary contains a key "name" and a key "nombre" 
del person['name'] # then delete the previous unneeded key

person['edad'] = person['age']
del person['age']

print(person)
```

### .get()
The .get() method is used to retrieve the value associated with a given key. It provides a way to access dictionary values safely, avoiding errors that occur when trying to access a non-existent key directly.

### .setdefault()
The .setdefault() method is used to **retrieve the value of a given key** in a dictionary. If the key does not exist, it **inserts the key** with a specified default value and returns that value.

```python
#Example: setdefault to retrieve values in a key
chicago_bulls_1991_1 = {23: "Michael Jordan", 33: "Scottie Pippen"}#create a new dict for this example

# Imagine we don't know if key=23 contains a player name and in case it doesn't we want to fill it with "unknown player"
player = chicago_bulls_1991_1.setdefault(23, "Unknown Player") #the key exists, so the new value is not added. It just retrieves the existing value
print("The content of key 23 is not changed: ",player)

print(chicago_bulls_1991_1)
# Output: {23: 'Michael Jordan', 33: 'Scottie Pippen'}
```

```python
#Example 2: Using .setdefault() to Add a New Key
player = chicago_bulls_1991_1.setdefault(91, "Dennis Rodman") #Despite saving a new object, this MODIFIES THE ORIGINAL DICTIONARY adding dennis rodman
print(player)  # Output: Dennis Rodman

print(chicago_bulls_1991_1)
```

```python
#Example 3: Difference Between .get() and .setdefault()
# .get() does NOT modify the dictionary
player = chicago_bulls_1991_1.get(10, "Unknown Player")
print(player)  # Output: Unknown Player
print(chicago_bulls_1991_1)
# Output: {23: 'Michael Jordan', 33: 'Scottie Pippen', 91: 'Dennis Rodman'}

# .setdefault() MODIFIES the dictionary
player = chicago_bulls_1991_1.setdefault(10, "Unknown Player")
print(player)  # Output: Unknown Player
print(chicago_bulls_1991_1)
```

# Loops in a dictionary

In Python, dictionaries are iterable objects, meaning you can loop through them. There are different ways to loop through a dictionary:
- by keys
- by values
- by key-value pairs.

## Loops over key-value pairs

We can use the for-in loop together with the function *.items()* method.

Syntax:
~~~
for *key*, *value* in dictionary.items():
    Access the key and value
    print(key, value)
~~~

```python
#define a dict
chicago_bulls_1991 = {
    0: "B.J. Armstrong",
    5: "John Paxson",
    10: "Stacey King"
}



# Loop through keys and values
for player_number, player_name in chicago_bulls_1991.items(): #I'm using a placeholder for both keys and values
    print(f"Player number {player_number} is {player_name}")
```

## Loops through all the keys in a dictionary

Python provides a clear syntax for looping through the keys of a dictionary:
~~~
for key in dictionary.keys():
    # Access the key
    print(key)
~~~


```python
chicago_bulls_1991 = {
    0: "B.J. Armstrong",
    5: "John Paxson",
    10: "Stacey King"
}

# Loop through keys
for player_number in chicago_bulls_1991.keys():
    print(player_number)
```

We can avoid the function *.keys()* and the result will be the same:

```python
for player_number in chicago_bulls_1991:
    print(player_number)
```

## Loops through all the values in a dictionary

Python also provides a simple syntax for iterating through all the values of a dictionary:

~~~
for key in dictionary.values():
    # Access the key
    print(key)
~~~


```python
chicago_bulls_1991 = {
    0: "B.J. Armstrong",
    5: "John Paxson",
    10: "Stacey King"
}

# Loop through values
for player_name in chicago_bulls_1991.values():
    print(player_name)
```

## An example: the password game
We can use the dictionary to create a more realistic example. Imagine we want to play the game 'Password': using the forbidden words the user has to guess the corresponding word.

```python
# We create a dictionary with the word to be guessed (key) and the forbidden words (value):
password_game = {'knife':"fork, cut, food",
                'car': "vehicule, wheels, truck",
                'dog': "animal, bite, bark"
                }
print(password_game)
```

Now we can expose the forbidden works to the user to guess the word:

```python
for word, forbidden in password_game.items(): #loops through the keys (word) and values (forbidden) and prints the forbiddens. Then asks user's input
    print(f"Forbidden words: {forbidden}")
    answer = input("What is the word? ").lower()#lower ensures the answer in case-insensitive

    if answer == word: #check if the inputed word coincides with  the iterated word
        print("Congratulations! You guessed it!")
    else:
        print("Sorry, that is not the word.")
```

A bookstore keeps track of its book inventory using a dictionary where:

The keys are book titles.
The values are the number of copies available.
The dictionary is structured as follows:

~~~
books = {
    "The Great Gatsby": 4,
    "To Kill a Mockingbird": 6,
    "1984": 3,
    "Pride and Prejudice": 8,
    "Moby-Dick": 2

Tasks:<br>
- Loop through the dictionary and print each book title with its available stock in the format:
"The Great Gatsby: 4 copies available"

- Calculate and print the total number of books in stock across all titles.

- List books that have fewer than 5 copies in stock.




```python
# Given book inventory
books = {
    "The Great Gatsby": 4,
    "To Kill a Mockingbird": 6,
    "1984": 3,
    "Pride and Prejudice": 8,
    "Moby-Dick": 2
}

# 1. Print each book title and its stock
print("Book Inventory:")
for title, stock in books.items(): #use .items to loop across key-value pairs
    print(f"{title}: {stock} copies available")

# 2. Calculate total stock
total_stock = 0
for stock in books.values():
    total_stock += stock

print(f"\nTotal books in stock: {total_stock}")

# 3. List books with fewer than 5 copies
print("\nBooks with stock below 5:")
for title, stock in books.items():
    if stock < 5:
        print(f"{title}: {stock} copies")
```

# Loops through a dictionary in order

Dictionaries are quite useful because they allow bits of information to be connected. One of the issues with dictionaries is that they are not stored in any particular order. When you retrieve all the keys or values in the dictionary, you can't be certain of the order in which you'll retrieve them. However, there is a quick and easy way to do this when you want them in a particular order using the *sorted()* function.

```python
word_definitions = {
    "computer": "An electronic device that processes data.",
    "book": "A set of written or printed pages fastened together in a cover.",
    "dog": "A domesticated carnivorous mammal.",
    "tree": "A woody perennial plant, typically having a single stem or trunk."
}

# Show rhe words in the dictionary without sorting:
print('Dictionary not sorted:')
for key in word_definitions.keys():
    print("- %s" % key)#prints in the format of a bulle points list

print('-'*20)
#Now sorting alphabetically
print('Dictionary sorted')
for key in sorted(word_definitions.keys()):
    print("- %s" % key)
```

### Iterating across nested dictionaries
A nested dictionary is a dictionary where values are themselves dictionaries.

For example, consider the following nested dictionary:

~~~
data = {
    "Category1": {"ObjectA": 10, "ObjectB": 20},
    "Category2": {"ObjectC": 15, "ObjectD": 25}
}
~~~
Here:<br>
- "Category1" and "Category2" are keys representing different categories.
- Each category contains another dictionary with items and their values.

To access values in a nested dictionary:
~~~
print(data["Category1"]["ObjectA"])  # Output: 10
~~~
To iterate over nested dictionaries:
~~~
for category, objects in data.items():
    for object, value in object.items():
        print(f"{object} has a value of {value}")
~~~
