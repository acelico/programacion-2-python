## 2. Core Data Structures

Pandas mainly uses two data structures: **Series** and **DataFrame**.

### Series

A **Series** is a one-dimensional array-like object that can hold various data types. It has an associated array of data labels called its index.

**Example: Creating a Series**

````python
# Creating a Series from a list
numbers = [10, 20, 30, 40, 50]
series_numbers = pd.Series(numbers)
print("Series:\n", series_numbers)
````

### DataFrames

A **DataFrame** is a two-dimensional, size-mutable, and heterogeneous tabular data structure with labeled axes (rows and columns). To make a comparison, one could think of an Excel spreadsheet as something similar (they're not the same but they resemble each other).

**Example: Creating a DataFrame**

````python
#Creating a dataframe from a dictionary
# 1: Create a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
# 2. Turn it into pandas DF
df = pd.DataFrame(data)
print("DataFrame:\n", df)
df
````

### Basic Pandas Attributes
When working with pandas, every DataFrame and Series comes with a number of attributes that help you understand its structure and content. Here are some essential ones:

- **index**:
Represents the labels for the rows. For a new DataFrame, the default index is a range starting at 0. You can customize it to use meaningful labels.

- **columns**:
Contains the column labels for a DataFrame. This is useful to quickly check which variables your data contains.

- **shape**:
Returns a tuple (number_of_rows, number_of_columns) indicating the dimensions of the DataFrame.

- **dtypes**:
Displays the data type of each column. This helps ensure that numerical data isn’t accidentally stored as strings, for example.

- **head**() and **tail**():
These methods let you preview the first or last few rows of your DataFrame, which is useful for a quick inspection of your data.

- **values**:
Provides the underlying data as a NumPy array if you need to work directly with the array.

# Data cleaning

## Missing values
When working with datasets in Python, missing values (NaNs) can cause issues in analysis and modeling. Here’s how you can handle them effectively using Pandas:
- Identifying missing values
- Removing missing values
- Filling missing values
- Interpolating missing values

### Identifying missing values
.isnull() method

````python
df_nba = pd.read_csv("nba.csv")
print(df_nba.isnull())  # Boolean mask (True if missing)
print(df_nba.isnull().sum())  # Count missing values per column
print(df_nba.info())  # Overview of missing values
````

## Removing missing values
Careful: when deciding to drop missing values we are dropping potentially valueable information.

````python
#Create a new dataset dropping columns with missing values from the original one
df_cleaned = df_nba.dropna(axis=1)
print(df_cleaned.shape)
#Create a new dataset dropping drop rows with missing values
df_cleaned = df_nba.dropna(axis=0)
print(df_cleaned.shape)
#This is NOT removing NAn from the original df
````

````python
print(df_nba.shape)
print(df_nba.dropna(how='any').shape) #We remove the rows that have any NA
df_nba.dropna(how='any').head() #Applying dropna, the rows are not removed from the original DataFrame
````

````python
df_nba.dropna(how='any',axis=0,inplace=True) # With inplace=True, the rows (axis=0) are removed from the original DataFrame
df_nba.head()
df_nba.shape
````

We can remove the rows that have NAs in a column using the parameter *subset*.

````python
df_nba = pd.read_csv("nba.csv")
print(df_nba.dropna(subset=['Salary']).shape) # We only remove the rows with NAs in Salary
df_nba.dropna(subset=['Salary']).head()
````

## Joins
(This is a very important topic for your assignment. You may want to do some research for more information.) <br>
Joins are essential in data manipulation and analysis when dealing with multiple datasets that share common elements.

**When Do We Need Joins?**
- Combining Related Datasets: <br>
Example: You have a Customers dataset and a Purchases dataset. A join helps link each customer with their respective purchases.

- Enriching Data with Additional Information:<br>
Example: You have a dataset of Students and another dataset of Test Scores. A join helps merge student details with their respective scores.

- Filtering Data Based on Matches/Mismatches:<br>
Example: You have an Employees dataset and a Salaries dataset. A join can help find employees with or without salary records.

- Avoiding Data Duplication & Redundancy:<br>
Instead of storing all information in one huge table, joins allow normalization, keeping data in separate, structured tables.

**An example** <br>
Let's use a toy dataset about employees and their salaries:
- We want to combine these two pieces of information

However, there are multiple ways to do so

````python
#Let's creae a toy dataset

# Creating first DataFrame (Employees)
df_employees = pd.DataFrame({
    'Employee_ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hank'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'Sales']
})

# Creating second DataFrame (Salaries)
df_salaries = pd.DataFrame({
    'Employee_ID': [3, 4, 5, 6, 7, 9, 10],
    'Salary': [60000, 75000, 50000, 80000, 65000, 70000, 72000]
})

print("Employees DataFrame:")
print(df_employees)

print("\nSalaries DataFrame:")
print(df_salaries)
````

1. INNER JOIN (Intersection)

It keeps exclusively records that exist in **both tables**. Observations that do not share the ID are left out.

````python
df_inner = pd.merge(df_employees, df_salaries, on='Employee_ID', how='inner')
print(df_inner)
````

2. LEFT JOIN (Keep All Left Table Records)<br>
It keeps all the observations specified in the left table. In practice, this means keeping all employees names adding their salary, if available.

````python
df_left = pd.merge(df_employees, df_salaries, on='Employee_ID', how='left')
print(df_left)
````

3. RIGHT JOIN (Keep All Right Table Records)<br>
It keeps all the observations specified in the right table. In practice, it keeps all salary records and adds employee details if available

4. FULL OUTER JOIN (Keep All Records)
Keeps all records from both tables. No observation is discarded. In practice,keeps all employees and all salary records

````python
df_outer = pd.merge(df_employees, df_salaries, on='Employee_ID', how='outer')
print(df_outer)
````
