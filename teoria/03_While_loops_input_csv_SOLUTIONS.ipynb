## Error Control

Error control in Python using try-except is a mechanism for handling exceptions, allowing a program to recover from unexpected issues without crashing. This structure helps in separating normal execution flow from error-handling logic. It basically allow the program getting to a complete execution also in presence of an error.

Syntax:
~~~
------------
try Block:
Contains code that could potentially throw an exception.
If no exception occurs, the except block is skipped.
------------

------------
except Block:
Catches specific or general exceptions.
If an exception occurs in the try block, control transfers to the except block, bypassing the remaining code in the try.
-----------
~~~

![](imagenes/try-except%20logic.png)

For instance, let's write a program that calculates the sum of the numbers introduced by a user, ensuring that the program does not crash if the user introduces anything different from a number. 

```python
a = input('Introduce the first integer number: ')
b = input('Introduce the second integer number: ')

try:
    a = int(a)
    print(f"Great, you introduced an integer: {a}")
except:
    print('Careful. The first number is not an integer.')

try: # si no ha surgido excepción va al siguiente bloque try
    b = int(b)
    print(f"Great, you introduced an integer: {b}")
except:
    print('Careful. The second number is not an integer.')

try:
    result = a + b
    print(f"The sum of {a} and {b} is: {result}")
except:
    print('The sum could not be calculated')
```

```python
# let's make it a bit more sophisticated:
# I want the program throw me a warning telling me what type of object i introduced when not integer
a = input('Introduce the first integer number: ')
b = input('Introduce the second integer number: ')

try:
    a = int(a)
    print(f"Great, you introduced an integer: {a}")
except:
    try:# we get here when a is not integer: it can still be float or string 
        float(a) #if is float this try works fine
        print("Careful. The first value is a float.")
    except: # if is not float the try skips to except and by exclusion we know a is string
        print("Careful. The first value is a string.")

try:
    b = int(b)
    print(f"Great, you introduced an integer: {b}")
except:
    try:
        float(b)
        print("Careful. The second value is a float.")
    except:
        print("Careful. The second value is a string.")

try:
    print(f'The sum (or concatenation) of {a} and {b} is: {a + b}')
except:
    print('The sum could not be calculated')
```

# While loop

The while loop in Python is used to execute a block of code repeatedly as long as a specified condition is **TRUE**. <br>

It is ideal for situations where the number of iterations isn't predetermined, and the loop depends on a condition to stop.<br>

Remeber: Indentation also applies to while loops!

Syntax:
~~~
------------------------------
 while condition:
      Code to execute repeatedly
      (Body of the loop)
-------------------------------
condition: An expression that evaluates to True or False.
The loop runs as long as the condition is True.
~~~



![](imagenes//while%20loop%20logic%20.png)

```python
#A program that prints all integers <= 3
i = 1
while i <= 5:
    print(i)
    i = i + 1 #this updates the initial variable i and allows the program keep running (i is 1, then becomes 2, then becomes 3, then the condition is not satisfied and loop interrupts)
print("Program finished") #stopping condition

#Key concepts
#Initialization: The starting point of the loop (num = 1).
#Condition: The loop continues running as long as the condition (num <= 3) is True.
#Incrementing: We change the value of num after each iteration (i = i + 1).
#Termination: The loop ends when the condition is no longer satisfied.
```

```python
# A program that prints all incremental numbers that multiplied by 3 are smaller than 50
i = 1
while i <= 50:
    print(i)
    i = 3*i + 1
print("Program finished")
```

**DANGER!!!!** One of the things you should be concerned about is that if the condition in the *while loop* is **always met**, then there will be an infinite loop, the loop will never end and then we will have to interrupt the kernel and re-start session again.

## Break statement

The break statement in Python is used to terminate the nearest enclosing loop prematurely, when a certain condition is met.<br>
It allows you to exit a for or while loop when a specific condition is met, and the control flow immediately continues with the first statement after the loop.

Syntax:
~~~
while condition:
    if condition:
        break

break: stops the execution of the loop and transfers control to the next line after the loop.
~~~

```python
while True: # This would call an infinite loop
    user_input = input("Enter 'stop' to quit: ") #we would be asked to input something to infinite
    if user_input == "stop": #if you enter sotp, "break" stops the loop and skips to the print statement
        break
    print("You entered:", user_input) #this is always printed as it is indented only under the "while"
print("Word 'stop' breaks the loop")
```

```python
# another example: write a program running until a number is smaller than 5 
numbers = [1, 2, 3, 4, 5, 6]

for num in numbers:
    if num > 5:  
        break
    print(f"{num} is smaller than or equal to 5")#this runs when num is < 5 

print(f"{num} is greater than 5")     
```

# Loops _while_ to have programs running

Most of the programs we are using everyday are executed until we indicate them to quit, and this is often done with a *while loop*.

Let's code a program in which the user can be entering numbers until he/she decides to enter a 0; then, we can then  calculate the sum of the elements introduced.<br> <br>
Here we will combine multiple different statements:
- While
- If
- Try - Except
- Break


```python
numbers = []         # Initialize the list to store numbers
total = 0            # Initialize a variable to track the sum

while True:  # infinite loop
    num = input("Write a number (or '0' to finish): ")  # user input

    if num == '0':  # exit condition
        break

    # Convert input to float and process
    try:
        num_float = float(num)
        numbers.append(num_float)  # Add the number to the list
        total += num_float         # Add the number to the sum
    except:  # Handle non-numeric input ( a type of error when we introdce a value that can't be processd) 
        print("Invalid input. Please enter a valid number.")

# Print results
print("Numbers entered:", numbers)
print("The sum of the sequence of numbers is:", total)
```

## Focus: the difference between *try-except* and *break* statements
Both try-except and break are control flow statements, but they serve  different purposes. <br> <br>
**Try-Except**:
- The try-except block is used for error handling.
- This means the program follows a different path when meets an error, and still gets to completion.
- The program does not crash in presence of errors.
<br>

**Break**:
- It has a different purpose: exiting a loop
- It immediately exits the loop when a certain condition is met.
- Does not handle errors or exceptions (the program can crash if error occurs)

# *Using while loops to process the elements in a list*
Last class we saw that we can apply *pop()* to the elements of a list. We can use a list of elements and process them one by one as we need it.


```python
# List of students to evaluate
students_not_ev = ['juan', 'maria', 'marta', 'fabián']#Students not yet evaluated
students_ev = []#an empty list for students evaluated

# We go through the elements of the list
while len(students_not_ev) > 0:

    # We evaluate a student and confirm the evaluation
    actual_student = students_not_ev.pop()
    print("Evaluating student: %s...evaluated!" % actual_student.title())

    # We append it to the list of students evaluated
    students_ev.append(actual_student)


print("\nStudents not evaluated:")
for student in students_not_ev:
    print('- ' + student.title())

print("\nStudents evaluated:")
for student in students_ev:
    print('- ' + student.title())
```

If we want the program to run as a FIFO, we can change the code... how?

```python
# List of students to evaluate
students_not_ev = ['juan', 'maria', 'marta', 'fabián']
students_ev = []

# We go through the elements of the list
while len(students_not_ev) > 0:

    # We evaluate a student and confirm the evaluation
    actual_student = students_not_ev.pop(0)
    print("Evaluating student: %s...evaluated!" % actual_student.title())

    # We append it to the list of students evaluated
    students_ev.append(actual_student)


print("\nStudents not evaluated:")
for student in students_not_ev:
    print('- ' + student.title())

print("\nStudents evaluated:")
for student in students_ev:
    print('- ' + student.title())
```

```python
#import a module to read and write csv files
import csv
csv_path = "PPR-2015.csv"
n_rows = 5 # the number of rows we want to print

#The context manager opens the file for reading in a variable file
with open(csv_path, "r") as file: #The with statement in Python is used to wrap the execution of a block of code within a context manager.
    reader = csv.reader(file) #reader is a function contained into the csv module. It creates a CSV reader object that will read the data from the file object. This object can then be used to iterate over the lines (rows) in the CSV file.
    # we can use a for-in loop
    for n, row in enumerate(reader):
        print(row)
        if n ==n_rows: # stop reading rows when you get to the indicated row number
            break

# You can see in the output that each row is contained in the form of a list. So this is not yet a dataframe! What we have is a set of lists (see the syntax with "[]")
```

If we want to read the whole file into a unique list:
- Create an empty list
- Read each row as a list
- Append each list to the previous to get the entire file into a single list

```python
import csv
csv_path = "PPR-2015.csv"
n_rows = 5
list_h=[]

#The context manager opens the file for reading in a variable file
with open(csv_path, "r") as file:
    reader = csv.reader(file)
    # we can use a for-in loop
    for n, row in enumerate(reader):
        list_h.append(row)
        # print(row)


print('There are %s offering flats and houses in Dublin' % (len(list_h)-1))
```

Now the variable *list_h* contains the csv file.
 In its first element, there is a list containing the file header with the column names, and from there, the values of each column. As you can see, it stores all the values as strings, so you will need to perform casting if you want to use another type, using the functions *int()* and *float()* for integer and floating-point types, respectively.

```python
print('This is the header ')
print(list_h[0])# each row of the dataset is now contained at specific positions of the list
print('1st row')
print(list_h[1])
```

One way of accessing the information of a column is using a for loop. Let's calculate the average price of the houses in the file.<br>

Notice from the previous output that "Price" is stored in position 4 in the list

```python

l_prices=[]
num_col=4 # column of the Price in euros
for row in range(1, len(list_h)):
    price=list_h[row][num_col] #We extract the price of the house, with the format '€138,000.00'
    price=price.split('.') #We split the string in two strings: ['€138,000','00']
    price=price[0]#We get the first part
    l_prices.append(int(price[1:].replace(',', ''))) #extract the characters starting in the second position so to eliminate the "€" and eliminate the',', and finally convert it into an integer

av_price=0 #set an initial average price
for price in l_prices:
  av_price+=price #add each new price to the initial average price
av_price=av_price/len(l_prices) #divide by the number of registered prices
print('The average price of the houses in Dublin is %f'%av_price,' euros')
```
